# **mycustom filter**

```python
# RAHA Filter
R = 1.4   # Brightness
A = 1.8   # Color
H = 1.3   # Contrast


```

```python
#filter mixer
edited = ImageEnhance.Brightness(photo).enhance(R)
edited = ImageEnhance.Color(edited).enhance(A)
edited = ImageEnhance.Contrast(edited).enhance(H)
edited = photo.filter(ImageFilter.SMOOTH_MORE)



```

```python
display(edited)
```

```python
#function
def raha_filter(strength=1.0):

    edited = ImageEnhance.Brightness(photo).enhance(R)
    edited = ImageEnhance.Color(edited).enhance(A)
    edited = ImageEnhance.Contrast(edited).enhance(H)
    edited = photo.filter(ImageFilter.SMOOTH_MORE)



```

```python
import ipywidgets as widgets
widgets.interact(
    raha_filter,
    strength=widgets.FloatSlider(
        min=0,
        max=2,
        step=0.1,
        value=1
    )
)
```
